```python
comparison_df = pd.DataFrame({
    "Model":["KNN","ANN","Random Forest","SVR","XGBoost"],
    "Train R2":[0.965,0.9821,0.9948,0.9906,0.998],
    "Test R2":[0.917,0.9664,0.9676,0.9867,0.9846],
    "MAE":[0.0687,0.0467,0.0382,0.02505,0.0312],
    "RMSE":[0.1653,0.1051,0.1032,0.0662,0.0712],
    "CV R2":[0.9206,0.9764,0.9576,0.9806,0.9799]
})

comparison_df
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

bars = plt.bar(
    comparison_df["Model"],
    comparison_df["Test R2"],
    color=[  # Linear
        "#3498db",  # KNN
        "#2ecc71",  # ANN
        "#f39c12",  # RF
        "#9b59b6",  # SVR
        "#e74c3c"   # XGB
    ],
    edgecolor="black",
    linewidth=1.2
)

# Zoomed axis
plt.ylim(0.90, 1.00)

plt.ylabel(
    r"$R^2$ Score",
    fontsize=13,
    fontweight="bold"
)

plt.xlabel(
    "Machine Learning Models",
    fontsize=13,
    fontweight="bold"
)

plt.title(
    "Comparison of Model Performance",
    fontsize=15,
    fontweight="bold"
)

# Grid
plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.6
)

# Value labels
for bar in bars:

    height = bar.get_height()

    plt.text(
        bar.get_x() + bar.get_width()/2,
        height + 0.001,
        f"{height:.4f}",
        ha="center",
        fontsize=10,
        fontweight="bold"
    )

plt.xticks(
    rotation=20,
    fontsize=11
)

plt.yticks(
    fontsize=11
)

plt.tight_layout()

plt.show()
```
